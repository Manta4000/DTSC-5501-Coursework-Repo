## Fundamentals of Quantum computing

### The State Vector

* the difference between the state of a classical computer and a Q computer is that the state of a vector in classical computation is deterministic
* while the state vector determines a probability distribution across a series of possible outputs
* the readout of a quantum computer dictates the *qubit number* of the program in question, where there are $2^k$ distinct possible outputs, any program provides a distribution across all of them, and the readout has $k$ distinct "bits"

### Definition: Qubit
    * a unit vector in a 2D space together w a coordinate system where two perpendicular x & y directions correspond to the two values you could read out once measurements are done
