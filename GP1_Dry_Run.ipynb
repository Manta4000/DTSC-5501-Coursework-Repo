### LinkedRequestList

* create a logical `LinkedList` with Nodes
* each node must have a nested `MaintainenceRequest`
* goal is to recreate the structure of a `LinkedList` with an interface attuned for the `MaintainenceRequest` class

|Methods|Req. Behavior|
|---|:--|
|`push(request)`|insert @ beginning|
|`append(request)`|insert @ end |
|`count()`|return no. of stored requests|
|`getAt(index)`|returns request @ a valid node/position|
|`insertAt(index, request)`|insert a request before current item at that position|
|`removeAt(index)`|remove & measure request @ that position|

|`__str__()`|return a readable of whole structure|
