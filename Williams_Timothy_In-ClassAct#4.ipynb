```python
## Constructing the tree 

class Node:
    def __init__(self, key, left=None, right=None):
        self.key = key
        self.left = left
        self.right = right


    def insert(node, key):
        if node is None:
            return Node(key)
        if key < node.key:
            node.left = insert(node.left, key)
        else:
            node.right = insert(node.right, key)
        return node

    def _display_aux(self):
    ### Returns list of strings, width, height, and horizontal coordinate of the root.
    # zero children
        if self.right is None and self.left is None:
            line = "%s" % self.key
            width = len(line)
            height = 1
            middle = width // 2
            return [line], width, height, middle
        # Only one children (starting with left)
        if self.right is None:
            lines, n, p, x = self.left._display_aux()
            s = "%s" % self.key
            u = len(s)
            first_line = (x + 1) * " " + (n - x - 1) * "_" + s
            second_line = x * " " + "/" + (n - x - 1 + u) * " "
            shifted_lines = [line + u * " " for line in lines]
            return [first_line, second_line] + shifted_lines, n + u, p + 2, n + u // 2
        # One childrend (starting with right)
        if self.left is None:
            lines, n, p, x = self.right._display_aux()
            s = "%s" % self.key
            u = len(s)
            first_line = s + x * "_" + (n - x) * " "
            second_line = (u + x) * " " + "\\" + (n - x - 1) * " "
            shifted_lines = [u * " " + line for line in lines]
            return [first_line, second_line] + shifted_lines, n + u, p + 2, u // 2
        # Second children.
        left, n, p, x = self.left._display_aux()
        right, m, q, y = self.right._display_aux()
        s = "%s" % self.key
        u = len(s)
        first_line = (x + 1) * " " + (n - x - 1) * "_" + s + y * "_" + (m - y) * " "
        second_line = (
            x * " " + "/" + (n - x - 1 + u + y) * " " + "\\" + (m - y - 1) * " "
        )
        if p < q:
            left += [n * " "] * (q - p)
        elif q < p:
            right += [m * " "] * (p - q)
        zipped_lines = zip(left, right)
        lines = [first_line, second_line] + [a + u * " " + b for a, b in zipped_lines]
        return lines, n + m + u, max(p, q) + 2, n + u // 2

    def PrintPreOrder(node):
        if node is not None:
            print(node.key, end=" ")
            Node.PrintPreOrder(node.left)
            Node.PrintPreOrder(node.right)
    
    def PrintInOrder(node):
        if node is not None:
            Node.PrintInOrder(node.left)
            print(node.key, end=" ")
            Node.PrintInOrder(node.right)
    
    def PrintPostOrder(node):
        if node is not None:
            Node.PrintPostOrder(node.left)
            Node.PrintPostOrder(node.right)
            print(node.key, end=" ")
```

```python
### Display the tree

def display(self):
        lines, *_ = self._display_aux()
        for line in lines:
            print(line)
```

```python
### Pt. 4 -- Inserting nodes into the BST
def insertBST(node, value):
    if node is None:
        return Node(value)
    if value < node.key:
        node.left = insertBST(node.left, value)
    else:
        node.right = insertBST(node.right, value)
    return node
```

```python
### Pt. 6 -- Rotating a BST Left and right

def rightRotation(node):
    newRoot = node.left
    node.left = newRoot.right
    newRoot.right = node
    return newRoot

def leftRotation(node):
    newRoot = node.right
    node.right = newRoot.left
    newRoot.left = node
    return newRoot
```

```python
### Pt. 7 -- Finding the height of a root

def height(node):
    if node is None:
        return -1  # height of an empty tree is -1
    else:
        left_height = height(node.left)
        right_height = height(node.right)
        return 1 + max(left_height, right_height)
    
```

```python
### Pt. 8 --- balancing an unbalanced BST
def createListFromTree(root, listNodes):
    if root is not None:
        createListFromTree(root.left, listNodes)
        listNodes.append(root.key)
        createListFromTree(root.right, listNodes)

def buildBSTRecursively(sorted_list, start, end):
    if start > end:
        return None
    mid = (start + end) // 2
    root = Node(sorted_list[mid])
    root.left = buildBSTRecursively(sorted_list, start, mid - 1)
    root.right = buildBSTRecursively(sorted_list, mid + 1, end)
    return root

def buildBST(sorted_list):
    if not sorted_list:
        return None
    mid = len(sorted_list) // 2
    root = Node(sorted_list[mid])
    root.left = buildBST(sorted_list[:mid])
    root.right = buildBST(sorted_list[mid + 1:])
    return root
```
