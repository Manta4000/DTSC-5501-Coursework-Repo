### Definition and Set-Theoretic and Tuple Representation

A Linked List (in this case a Singly linked list) is defined as 'a collection of nodes that collectively form a linear sequence.' - Data Structures and Algorithms in Python (Goodrich, pp. 256 - 257)

Let $N = \{v_0, v_1, v_2, v_3, \dots, v_{n-1}\}$ be the set of nodes.
Thus each node $v_i$ is an ordered pair with the structure as follows:

$v_i = (d_i, next_i)$

where $d_i$ is the data from any given data set and $next_i$ represents a pointer to the subsequent node $v_{i+1}$

## Order of methods to be applied to class LinkedList and their behavior

1) __Push()__ $\rightarrow$ inserted at the begninning, including into an empty list
2) __append()__ $\rightarrow$ insert a value at the end, including into an empty list
3) __ __str__ __ __()__ $\rightarrow$ Build the chain text without changing the list
4) __count()__ $\rightarrow$ the amount of nodes are counted and then stored
5) __getAt()__ $\rightarrow$ read the value at a valid position
6) __insertAt()__ $\rightarrow$ inserts a value vefore the current position which allows an insertion at the end, where valid positions are defined within the following space;

$ 0 \leq index \leq n $

7) __removeAt()__ $\rightarrow$ removes any item in the list as long as it is at a valid position where;

$ 0 \leq index < n$

```python
"DTSC 5501 - In-Class Activity 3 - Linked Lists - Timothy Williams"

class Node:

    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:

    def __init__(self):
        self.head = None
        
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
        return True

    def append(self, value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node
        return True

    def __str__(self):
        values = []
        current = self.head
        while current:
            values.append(str(current.data))
            current = current.next
        values.append("None")
        return " → ".join(values)

    def count(self):
        current = self.head
        count = 0
        while current:
            count += 1
            current = current.next
        return count

    def getAt(self,index):
        if index < 0:
            raise IndexError("Index out of bounds")
        current = self.head
        for _ in range(index):
            if not current:
                raise IndexError("Index out of bounds")
            current = current.next
        if not current:
            raise IndexError("Index out of bounds")
        return current.data

    def insertAt(self, index, data):
        if index < 0 or index > self.count():
            raise IndexError("Index out of bounds")
        if index == 0:
            self.push(data)
            return
        current = self.head
        for _ in range(index - 1):
            current = current.next
        new_node = Node(data)
        new_node.next = current.next
        current.next = new_node
        return None

    def removeAt(self, index):
        if index < 0 or index >= self.count():
            raise IndexError("Index out of bounds")
        if index == 0:
            removed_node = self.head
            self.head = self.head.next
        else:
            current = self.head
            for _ in range(index - 1):
                current = current.next
            removed_node = current.next
            current.next = removed_node.next
        return removed_node.data

    def __len__(self):
        return self.count()

    def display(self):
        print(self)
```

```python
def check_empty():
    ll = LinkedList()
    assert ll.head is None
    assert ll.count() == 0
    assert len(ll) == 0
    assert str(ll) == "None"
    return True


def check_front_and_end():
    ll = LinkedList()
    assert ll.append(20)
    assert ll.append(40)
    assert ll.push(10)
    assert str(ll) == "10 → 20 → 40 → None"
    assert ll.count() == 3
    assert len(ll) == 3
    return True


def check_indexed_operations():
    ll = LinkedList()
    ll.append(20)
    ll.append(40)
    ll.push(10)
    assert ll.insertAt(2, 30) is None
    assert ll.getAt(2) == 30
    assert ll.removeAt(1) == 20
    assert str(ll) == "10 → 30 → 40 → None"
    assert ll.getAt(1) == 30
    assert ll.count() == 3
    return True
```

### Custom edge-case tests

1) **Pushing into an empty list**: Push one value and verify it becomes the head, has no next node, and the list length and string are correct.

2) **Removing the head, tail, and final item**: Remove the head, then the tail, then the only remaining node; verify the list after each operation, including when it becomes empty.

3) **Inserting and removing at the first and last positions**: Insert at index `0` and at index `len(ll)`, then remove both inserted values from their respective ends.

4) **Invalid indexing**: Check that negative and out-of-range indexes raise `IndexError` for `getAt`, `insertAt`, and `removeAt`.

5) **Using both numbers and strings**: Store integer, string, and floating-point values; verify retrieval preserves their original types and string formatting.

6) **Keeping a list unchanged after an operation fails**: Trigger invalid insertions and removal, then verify the list contents and length have not changed.

The code cell below runs these six groups and reports the number that pass.

```python
def expect_index_error(operation):
    try:
        operation()
    except IndexError:
        return
    raise AssertionError("Expected IndexError")


def check_push_into_empty_list():
    ll = LinkedList()
    ll.push(7)
    assert ll.head.data == 7
    assert ll.head.next is None
    assert str(ll) == "7 → None"
    assert len(ll) == 1


def check_remove_head_tail_and_final_item():
    ll = LinkedList()
    for value in (10, 20, 30):
        ll.append(value)

    assert ll.removeAt(0) == 10
    assert str(ll) == "20 → 30 → None"
    assert ll.removeAt(len(ll) - 1) == 30
    assert str(ll) == "20 → None"
    assert ll.removeAt(0) == 20
    assert ll.head is None
    assert str(ll) == "None"


def check_insert_and_remove_first_and_last():
    ll = LinkedList()
    ll.append("middle")
    ll.insertAt(0, "first")
    ll.insertAt(len(ll), "last")
    assert str(ll) == "first → middle → last → None"
    assert ll.removeAt(0) == "first"
    assert ll.removeAt(len(ll) - 1) == "last"
    assert str(ll) == "middle → None"


def check_invalid_indexing():
    ll = LinkedList()
    ll.append("only")
    expect_index_error(lambda: ll.getAt(-1))
    expect_index_error(lambda: ll.getAt(1))
    expect_index_error(lambda: ll.insertAt(-1, "invalid"))
    expect_index_error(lambda: ll.insertAt(2, "invalid"))
    expect_index_error(lambda: ll.removeAt(-1))
    expect_index_error(lambda: ll.removeAt(1))


def check_numbers_and_strings():
    ll = LinkedList()
    ll.append(42)
    ll.append("answer")
    ll.append(3.5)
    assert ll.getAt(0) == 42
    assert ll.getAt(1) == "answer"
    assert ll.getAt(2) == 3.5
    assert str(ll) == "42 → answer → 3.5 → None"


def check_failed_operation_preserves_list():
    ll = LinkedList()
    for value in (1, "two", 3):
        ll.append(value)
    before = str(ll)
    before_count = len(ll)

    for operation in (
        lambda: ll.insertAt(-1, "bad"),
        lambda: ll.insertAt(len(ll) + 1, "bad"),
        lambda: ll.removeAt(len(ll)),
    ):
        expect_index_error(operation)
        assert str(ll) == before
        assert len(ll) == before_count


edge_case_checks = [
    ("Push into empty list", check_push_into_empty_list),
    ("Remove head, tail, and final item", check_remove_head_tail_and_final_item),
    ("Insert/remove at first and last positions", check_insert_and_remove_first_and_last),
    ("Invalid indexing", check_invalid_indexing),
    ("Numbers and strings", check_numbers_and_strings),
    ("Failed operation preserves list", check_failed_operation_preserves_list),
]

edge_cases_passed = sum(run_check(name, check) for name, check in edge_case_checks)
print(f"Custom edge-case checks passed: {edge_cases_passed} of {len(edge_case_checks)}")
```

### Gonna make a Fibonacci sequence from 1st to 25th element

Then I will check the accuracy of the elements called by index

```python
def fib_seq(n):
    ll = LinkedList()
    x, y = 0, 1
    for _ in range(n):
        ll.append(x)
        x, y = y, x+y
    return ll

ll = fib_seq(10)
print("Fibonacci sequence with 10 elements: ", ll)
ll = fib_seq(25)
print("Fibonacci sequence with 25 elements: ", ll)

def check_fib(ll, n):
    for i in range(n):
        if ll.getAt(i) != ll.getAt(i):
            return False
    return True

print("Check the Fibonacci sequence from 0 to 9th element: " ,check_fib(ll, 10), " The 10th element is: ", ll.getAt(9))
```

### Time and space complexity

This implementation stores only `head`. It does not store a `tail` pointer or a cached node count. Consequently, operations that need the tail or validate an index by calling `count()` must traverse the list.

In the table, **locating/validation** is the cost to walk nodes or determine whether an index is valid; **link work** is the cost to change pointers after the target is known. Link changes themselves take constant time. Let $n$ be the number of nodes and $i$ the requested index.

| Operation | Locating or validation | Link work | Worst-case total time |
|---|---:|---:|---:|
| `push(data)` | None | $O(1)$ | $O(1)$ |
| `append(value)` | Walk to the last node: $O(n)$ | $O(1)$ | $O(n)$; empty-list case is $O(1)$ |
| `count()` | Walk all nodes: $O(n)$ | None | $O(n)$; empty-list case is $O(1)$ |
| `len(ll)` | Calls `count()`: $O(n)$ | None | $O(n)$; empty-list case is $O(1)$ |
| `getAt(i)` | Walk from the head to index $i$: $O(i)$ when valid, up to $O(n)$; negative index is rejected in $O(1)$ | None | $O(n)$; index `0` is $O(1)$ |
| `insertAt(i, data)` | Calls `count()` for validation, then may walk to index $i-1$: up to $O(n)$ total (two traversals) | $O(1)$ | $O(n)$; even index `0` is $O(n)$ for a nonempty list because validation counts first; empty-list insertion is $O(1)$ |
| `removeAt(i)` | Calls `count()` for validation, then may walk to index $i-1$: up to $O(n)$ total (two traversals) | $O(1)$ | $O(n)$; even removing index `0` is $O(n)$ for a nonempty list because validation counts first |
| `str(ll)` | Visit each node and convert its value to text | No list-link changes | $O(n + L)$ under the conversion assumption below |
| `display()` | Calls `str(ll)` and prints the result | No list-link changes | $O(n + L)$, excluding terminal/device-specific printing cost |

The index-operation table separates node traversal from pointer rewiring: insertion/removal rewires only a constant number of links once the location is reached. However, `insertAt` and `removeAt` first call `count()`, so their whole-method cost remains $O(n)$, including at the head. `append` also has no constant-time tail boundary except when the list is empty.

**Space.** The persistent list occupies $O(n)$ space for its nodes and their `data`/`next` references, plus $O(1)$ for the `LinkedList` object’s `head` reference. There is no persistent tail or count field. `push` and `append` add one persistent node ($O(1)$ additional space per call); `insertAt` does the same. `removeAt`, `count`, `len`, `getAt`, and index validation use $O(1)$ temporary working space. Returned data from `getAt`/`removeAt` is a reference to the stored value, not a copied value (assuming ordinary Python object-reference semantics).

For `str(ll)`, the method creates a temporary `values` list with $O(n)$ references and temporary text for converted values. The joined chain string is the returned output and occupies $O(L)$ space. Thus `str(ll)` has $O(n + L)$ temporary/output space in total; counting only auxiliary space and excluding the returned string, it still uses $O(n + L)$ because it holds the component strings while joining.

**Text-conversion assumption.** $L$ is the length of the complete returned chain string, including separators and the final `None`. The $O(n + L)$ string-time bound assumes each stored value’s `str(value)` takes time proportional to the number of characters it produces (and that values do not perform unusually expensive or side-effecting work in `__str__`). More generally, add the total cost of all value conversions to the $O(n + L)$ traversal and join work.
