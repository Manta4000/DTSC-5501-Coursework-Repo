### Definition and Set-Theoretic and Tuple Representation

A Linked List (in this case a Singly linked list) is defined as 'a collection of nodes that collectively form a linear sequence.' - Data Structures and Algorithms in Python (Goodrich, pp. 256 - 257)

Let $N = \{v_0, v_1, v_2, v_3, \dots, v_{n-1}\}$ be the set of nodes.
Thus each node $v_i$ is an ordered pair with the structure as follows:

$v_i = (d_i, next_i)$

where $d_i$ is the data from any given data set and $next_i$ represents a pointer to the subsequent node $v_{i+1}$

```python
"DTSC 5501 - In-Class Activity 3 - Linked Lists - Timothy Williams"

class Node:

    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:

    def __init__(self):
        self.head = None
        
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def append(self, value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node

    def __str__(self):
        values = []
        current = self.head
        while current:
            values.append(current.data)
            current = current.next
        return " -> ".join(map(str, values))

    def count(self):
        current = self.head
        count = 0
        while current:
            count += 1
            current = current.next
        return count

    def getAt(self,index):
        current = self.head
        for _ in range(index):
            if not current:
                raise IndexError("Index out of bounds")
            current = current.next
        if not current:
            raise IndexError("Index out of bounds")
        return current.data

    def insertAt(self, index, data):
        if index < 0 or index > self.count():
            raise IndexError("Index out of bounds")
        if index == 0:
            self.push(data)
            return
        current = self.head
        for _ in range(index - 1):
            current = current.next
        new_node = Node(data)
        new_node.next = current.next
        current.next = new_node

    def removeAt(self, index):
        if index < 0 or index >= self.count():
            raise IndexError("Index our of bounds")
        if index == 0:
            self.head = self.head.next
        else:
            current = self.head
            for _ in range(index - 1):
                current = current.next
            current.next = current.next.next
```

```python
ll = LinkedList()

ll.push(5)
ll.push(10)
ll.push(15)

print(ll)
```

```python
def check_empty():
    ll = LinkedList()
    assert ll.head is None
    assert ll.count() == 0
    assert len(ll) == 0
    assert str(ll) == "None"
    return True


def check_front_and_end():
    ll = LinkedList()
    assert ll.append(20) is None
    assert ll.append(40) is None
    assert ll.push(10) is None
    assert str(ll) == "10 → 20 → 40 → None"
    assert ll.count() == 3
    assert len(ll) == 3
    return True


def check_indexed_operations():
    ll = LinkedList()
    ll.append(20)
    ll.append(40)
    ll.push(10)
    assert ll.insertAt(2, 30) is None
    assert ll.getAt(2) == 30
    assert ll.removeAt(1) == 20
    assert str(ll) == "10 → 30 → 40 → None"
    assert ll.getAt(1) == 30
    assert ll.count() == 3
    return True
```

```python
def run_check(name, check):
    """Report unfinished work separately from a test failure."""
    try:
        check()
    except NotImplementedError as exc:
        print(f"NOT READY | {name}: {exc}")
        return False
    except Exception as exc:
        print(f"FAIL      | {name}: {type(exc).__name__}: {exc}")
        return False
    else:
        print(f"PASS      | {name}")
        return True


checks = [
    ("Empty list", check_empty),
    ("Beginning and end", check_front_and_end),
    ("Indexed operations", check_indexed_operations),
]
passed = sum(run_check(name, check) for name, check in checks)
print(f"Provided checks passed: {passed} of {len(checks)}")
print("The provided checks are not a complete test suite. Add your own below.")
```
